```python
import csv

data = [
    ['ID', 'Name', 'City', 'Salary'],
    ['1', 'Alice', 'New York, NY', '70000'],
    ['2', 'Bob', 'Los Angeles', '65000'],
    ['3', 'Charlie', 'Chicago', '80000']
]

with open(r'File Handling\employees.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(data)
```

# Example 1
1. The Standard Read Write a script that opens employees.csv and prints every row as a List of Strings.

Expected Output: ['1', 'Alice', 'New York, NY', '70000']

```python
with open(r'File Handling\employees.csv', 'r', newline='') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```
