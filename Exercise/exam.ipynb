# Q2: Write a Python function to count the frequency of each character in a
given string and return the output in a dictionary. Example:
char_frequency("HELLO") returns {"H":1, "E":1, "L":2, "O":1} 

```python
def counter(s):
    freq = {}
    for char in s:
        if char in freq:
            freq[char]+=1
        else:
            freq[char] = 1
    return freq
print(counter('hello'))
```

# Q3
Write a program to reverse the contents of a file character by character,
separating each character with a comma. 

```python
with open(r'File Handling\sample.txt','r+') as file:
    content = file.read()
    content=content[::-1]
    content = ",".join(content)
    file.seek(0)
    file.write(content)
```

# Q3
Write a program to validate email addresses using regular expressions.
Criteria:
1. Must contain @ symbol
2. Must contain domain name
3. Should not have spaces

```python
import re
s='sdfnjjn213@gmail.com'
pattern=r'^[\w\.\_]+@[\w\.\_]+\.[a-zA-Z]+$'
if re.match(pattern,s):
    print(True)
```
